# Reproduce the science report

The same English-language Python pipeline is used here and in the command line. Default settings are 2,000 bootstrap draws and 999 null draws. Individual response data and local diagnostics are not uploaded by this notebook.


````python
from pathlib import Path
import os, sys, subprocess

ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "run_pipeline.py").is_file()), None)
if ROOT is None:
    subprocess.run(["git", "clone", "https://github.com/HoangGiaBao107/AI-Use-Academic-Improvement-and-Unaided-Self-Study-among-Specialized-High-School-Students.git", "ai-learning-study"], check=True)
    ROOT = Path("ai-learning-study").resolve()
os.chdir(ROOT)
print(ROOT)
````

## Install dependencies

Run this before importing analysis libraries. If a hosted notebook asks for a runtime restart after installation, restart and rerun the setup cell.


````python
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)
````

## Provide the original CSV

Copy `data raw eng.csv` into `data/`. In Google Colab, the cell below opens the file uploader when the input is missing.


````python
DATA = ROOT / "data" / "data raw eng.csv"
if not DATA.is_file():
    try:
        from google.colab import files
        previous = Path.cwd()
        os.chdir(ROOT / "data")
        try:
            uploaded = files.upload()
        finally:
            os.chdir(previous)
    except ImportError:
        raise FileNotFoundError("Place data raw eng.csv in data/ before running the phases.")
if not DATA.is_file():
    raise FileNotFoundError("The input must be named data raw eng.csv.")
````

## Phase 1: Data And Codebook

This phase writes CSV tables and a phase-level ZIP. Tables are displayed below.


````python
subprocess.run([sys.executable, "run_pipeline.py", "--data", str(DATA), "--phases", "1", "--workers", "2"], check=True)
import pandas as pd
from IPython.display import display, Markdown
for table_path in sorted((ROOT / "results" / "tables" / "01_data").glob("*.csv")):
    display(Markdown("### " + table_path.stem))
    display(pd.read_csv(table_path))
````
